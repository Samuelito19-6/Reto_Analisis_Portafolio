## 1. Datos del portafolio

El portafolio está compuesto por cuatro activos. Cada activo tiene una cantidad de unidades y un precio de compra.

La información está organizada mediante un **diccionario** porque necesitamos asociar cada activo con sus datos.

### Datos de trabajo

- NVDA: 18 acciones a $284

- TSLA: 10 acciones a $192

- NTFLX: 12 acciones a $352

- AMZN: 7 acciones a $382


```python
# Diccionario con la información del portafolio

portafolio = {
   "NVDA": {"cantidad": 18, "precio_compra": 284},
    "TSLA": {"cantidad": 10, "precio_compra": 192},
    "NFLX": {"cantidad": 12, "precio_compra": 352},
    "AMZN": {"cantidad": 7, "precio_compra": 382}
}

# Lista de activos
activos = list(portafolio.keys())

print("Activos:", activos)
print("Información de NVDA:", portafolio["NVDA"])
```

## 2. Precios actuales

Ahora la firma recibe los precios actuales de mercado.

Para esta actividad trabajaremos con datos simulados para concentrarnos en la programación.

> Más adelante este mismo tipo de estructura podrá conectarse con datos reales.


```python
precios_actuales = {
    "NVDA": 354,
    "TSLA": 194,
    "NFLX": 420,
    "AMZN": 340
}

print(precios_actuales)
```

## 3. Valorar cada posición

Para cada activo necesitamos calcular:

**Valor invertido**

\[
Cantidad \times Precio\ de\ compra
\]

**Valor actual**

\[
Cantidad \times Precio\ actual
\]

Primero vamos a hacerlo para un solo activo. Después automatizaremos el proceso con un ciclo.


```python

cantidad = portafolio["NVDA"]["cantidad"]
precio_compra = portafolio["NVDA"]["precio_compra"]
precio_actual = precios_actuales["NVDA"]

valor_invertido = cantidad * precio_compra
valor_actual = cantidad * precio_actual

print("Valor invertido:", valor_invertido)
print("Valor actual:", valor_actual)
```

## 4. Automatizar con un ciclo `for`

Tenemos cuatro activos. Repetir el mismo código para cada uno sería poco eficiente.

Usaremos un ciclo `for` para recorrer la lista de activos y calcular los valores automáticamente.

### Tu turno

Completa la celda. El resultado debe mostrar el valor invertido y el valor actual de cada activo.


```python
# Completa el código

for activo in activos:
    cantidad = portafolio[activo]["cantidad"]
    precio_compra = portafolio[activo]["precio_compra"]
    precio_actual = precios_actuales[activo]

    valor_invertido = cantidad * precio_compra
    valor_actual = cantidad * precio_actual

    print(activo, "| Invertido:", valor_invertido, "| Actual:", valor_actual)
```

## 5. Ganancia, pérdida y rendimiento

Ahora agregaremos dos indicadores financieros:

**Ganancia o pérdida**

\[
Ganancia = Valor\ actual - Valor\ invertido
\]

**Rendimiento**

\[
Rendimiento = \frac{Ganancia}{Valor\ invertido}\times100
\]

Primero calcula estos indicadores para NVDA.


```python
ganancia = valor_actual - valor_invertido
rendimiento = (ganancia / valor_invertido) * 100

print("Ganancia / pérdida:", ganancia)
print("Rendimiento:", round(rendimiento, 2), "%")
```

## 6. Condicionales: interpretar el rendimiento

La firma utiliza una regla sencilla para generar una primera recomendación:

- **10 % o más:** MANTENER
- **Entre 0 % y 9.99 %:** REVISAR
- **Menor que 0 %:** ANALIZAR VENTA

Ahora el programa debe tomar una decisión utilizando `if`, `elif` y `else`.


```python
if rendimiento >= 10:
    recomendacion = "MANTENER"
elif rendimiento >= 0:
    recomendacion = "REVISAR"
else:
    recomendacion = "ANALIZAR VENTA"

print("Recomendación:", recomendacion)
```

## 7. Crear una función

Nuestro programa necesita realizar el mismo cálculo muchas veces.

En lugar de repetir instrucciones, crearemos una función que reciba la cantidad y los precios y devuelva los principales resultados.

### Tu turno

Completa la función.

La función debe devolver:

1. valor invertido;
2. valor actual;
3. ganancia o pérdida;
4. rendimiento.


```python
def analizar_activo(cantidad, precio_compra, precio_actual):
    valor_invertido = cantidad * precio_compra
    valor_actual = cantidad * precio_actual
    ganancia = valor_actual - valor_invertido
    rendimiento = (ganancia / valor_invertido) * 100

    return valor_invertido, valor_actual, ganancia, rendimiento


resultado = analizar_activo(28, 284, 354)
print(resultado)
```

## 8. Integrar todo el programa

Ahora construiremos el análisis completo del portafolio.

El programa debe:

1. recorrer todos los activos;
2. obtener sus datos;
3. utilizar la función `analizar_activo()`;
4. determinar la recomendación;
5. almacenar los resultados.

Para almacenar los resultados utilizaremos una **lista de diccionarios**.


```python
resultados = []

for activo in activos:
    cantidad = portafolio[activo]["cantidad"]
    precio_compra = portafolio[activo]["precio_compra"]
    precio_actual = precios_actuales[activo]

    valor_invertido, valor_actual, ganancia, rendimiento = analizar_activo(
        cantidad, precio_compra, precio_actual
    )

    if rendimiento >= 10:
        recomendacion = "MANTENER"
    elif rendimiento >= 0:
        recomendacion = "REVISAR"
    else:
        recomendacion = "ANALIZAR VENTA"

    resultados.append({
        "activo": activo,
        "valor_invertido": valor_invertido,
        "valor_actual": valor_actual,
        "ganancia": ganancia,
        "rendimiento": rendimiento,
        "recomendacion": recomendacion
    })

resultados
```

## 9. Informe del portafolio

Ahora utilizaremos los resultados para construir una salida más clara para el analista.

También calcularemos el valor total invertido, el valor total actual y la ganancia o pérdida total.


```python
total_invertido = 0
total_actual = 0

print("=" * 65)
print("ANÁLISIS DEL PORTAFOLIO")
print("=" * 65)

for resultado in resultados:
    total_invertido += resultado["valor_invertido"]
    total_actual += resultado["valor_actual"]

    print(
        f"{resultado['activo']:6} | "
        f"Actual: ${resultado['valor_actual']:,.0f} | "
        f"Rendimiento: {resultado['rendimiento']:.2f}% | "
        f"{resultado['recomendacion']}"
    )

ganancia_total = total_actual - total_invertido
rendimiento_total = (ganancia_total / total_invertido) * 100

print("=" * 65)
print(f"Total invertido: ${total_invertido:,.0f}")
print(f"Valor actual:    ${total_actual:,.0f}")
print(f"Ganancia/pérdida: ${ganancia_total:,.0f}")
print(f"Rendimiento:     {rendimiento_total:.2f}%")
```

## 10. Identificar el mejor activo

El programa ya conoce el rendimiento de cada activo.

Ahora utilizaremos una operación sobre los resultados para identificar cuál presenta el mayor rendimiento.



```python
mejor_activo = max(resultados, key=lambda x: x["rendimiento"])

print("Activo con mejor rendimiento:", mejor_activo["activo"])
print("Rendimiento:", round(mejor_activo["rendimiento"], 2), "%")
print("Recomendación:", mejor_activo["recomendacion"])
```
